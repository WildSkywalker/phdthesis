```python
from sklearn.datasets import fetch_openml
from sklearn.decomposition import PCA
import pandas as pd
import numpy as np
```

```python
mnist = fetch_openml("mnist_784", version=1, as_frame=False)
mnist.target = mnist.target.astype(np.uint8)

X_mnist = pd.DataFrame(mnist["data"])
y = pd.DataFrame(mnist["target"])

X_mnist = X_mnist.sample(n=1000)
y_mnist = y.loc[X_mnist.index]

X = PCA(n_components=30).fit_transform(X_mnist)
```

```python
pd.DataFrame(X)
```

```python
import general_tsne

X = np.array(X)

p_ij = general_tsne.get_original_pairwise_affinities(X)
```

```python
pd.DataFrame(p_ij)
```

```python
p_ij_symmetric = general_tsne.get_symmetric_p_ij(p_ij)
```

```python
pd.DataFrame(p_ij_symmetric)
```

```python
y0 = general_tsne.initialization(X, 2, "random")
```

```python
pd.DataFrame(y0)
```

```python
q_ij = general_tsne.get_low_dimensional_affinities(y0)
```

```python
pd.DataFrame(q_ij)
```

```python

gradient = general_tsne.get_gradient(p_ij_symmetric, q_ij, y0)
```

```python

pd.DataFrame(gradient)
```

```python
solution, Y = general_tsne.tsne(
    X, perplexity=10, T=1000, η=200, early_exaggeration=4, n_dimensions=2
)
```

```python

pd.DataFrame(solution)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
g1 = ax.scatter(solution[:, 0], solution[:, 1], c=y_mnist, cmap="tab10")
ax.axis("off")
ax.set_title("MNIST t-SNE")
plt.colorbar(g1, ax=ax)
plt.show()
```

```python
import plotly.express as px

fig = px.scatter(
    x=solution[:, 0],
    y=solution[:, 1],
    color=y_mnist[0].astype(str),  # Access the 'target' column and convert to string
    color_discrete_sequence=px.colors.qualitative.Set1,
    title="MNIST t-SNE Visualization",
    labels={"color": "Digit"},
    category_orders={"color": [str(i) for i in range(10)]},
)

fig.update_traces(marker=dict(size=10))

# Customize the layout
fig.update_xaxes(
    showticklabels=False,
    showgrid=False,
    zeroline=False,
    title="",  # Remove x-axis title
)
fig.update_yaxes(
    showticklabels=False,
    showgrid=False,
    zeroline=False,
    title="",  # Remove y-axis title
)
fig.update_layout(
    title_x=0.5,
    title_font = dict(size = 30), 
    plot_bgcolor="white",
    width = 800,
    height = 800,
    legend_title_font=dict(size=20),
    legend_font = dict(size = 20)
)

# Show the plot
fig.show()
```
